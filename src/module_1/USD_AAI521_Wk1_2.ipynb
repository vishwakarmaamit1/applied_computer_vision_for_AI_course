```python
# ================================================================
# AAI-521: APPLIED COMPUTER VISION FOR AI
# Module 1 Case Study:
# Automated Quality Inspection of Manufactured Products
# (Educational Image Classification Demo)
#
# TensorFlow / Keras CNN Implementation
#
# Dataset: Fashion-MNIST
# Libraries:
#   TensorFlow / Keras
#   Pillow
#   NumPy
#   Matplotlib
#   scikit-learn
#
# IMPORTANT:
# Fashion-MNIST supports clothing-category classification.
# It does NOT contain defective/non-defective manufacturing labels.
# This case study is an educational proxy for computer vision.
# ================================================================


# ================================================================
# STEP 0: IMPORT LIBRARIES
# ================================================================

import os
import random
import numpy as np
import matplotlib.pyplot as plt
import PIL
import sklearn
import tensorflow as tf

from PIL import Image, ImageEnhance, ImageFilter
from tensorflow import keras
from tensorflow.keras import layers

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)


# ================================================================
# STEP 0A: REPRODUCIBILITY AND ENVIRONMENT
# ================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("=" * 70)
print("AAI-521: APPLIED COMPUTER VISION FOR AI")
print("Module 1: CNN-Based Product Image Classification")
print("=" * 70)

print("TensorFlow version:", tf.__version__)
print("Keras version:", keras.__version__)
print("Pillow version:", PIL.__version__)
print("scikit-learn version:", sklearn.__version__)

gpus = tf.config.list_physical_devices("GPU")

if gpus:
    print("GPU available:", gpus[0])
else:
    print("GPU not available. Running on CPU.")

print()


# ================================================================
# CONFIGURATION
# ================================================================

OUTPUT_DIR = "aai521_cnn_outputs"

BATCH_SIZE = 64
EPOCHS = 5
LEARNING_RATE = 0.001

os.makedirs(OUTPUT_DIR, exist_ok=True)

print("Configuration")
print("-" * 70)
print("Batch size:", BATCH_SIZE)
print("Epochs:", EPOCHS)
print("Learning rate:", LEARNING_RATE)
print("Output directory:", OUTPUT_DIR)
print()


# ================================================================
# STEP 1: LOAD AND UNDERSTAND THE DATASET
# ================================================================

print("=" * 70)
print("STEP 1: LOAD AND UNDERSTAND THE DATASET")
print("=" * 70)

(x_train_full, y_train_full), (x_test_raw, y_test) = (
    keras.datasets.fashion_mnist.load_data()
)

print("Original training dataset:", x_train_full.shape)
print("Independent test dataset:", x_test_raw.shape)
print("Training labels shape:", y_train_full.shape)
print("Test labels shape:", y_test.shape)
print("Image shape:", x_train_full[0].shape)
print("Minimum pixel value:", x_train_full.min())
print("Maximum pixel value:", x_train_full.max())
print()


# ================================================================
# FASHION-MNIST CLASS NAMES
# ================================================================

class_names = [
    "T-shirt/top",
    "Trouser",
    "Pullover",
    "Dress",
    "Coat",
    "Sandal",
    "Shirt",
    "Sneaker",
    "Bag",
    "Ankle boot"
]

NUM_CLASSES = len(class_names)

print("Number of classes:", NUM_CLASSES)

for i, class_name in enumerate(class_names):
    print(i, ":", class_name)

print()


# ================================================================
# STEP 1A: VISUALIZE SAMPLE IMAGES
# ================================================================

print("=" * 70)
print("STEP 1A: VISUALIZE SAMPLE IMAGES")
print("=" * 70)

fig, axes = plt.subplots(2, 5, figsize=(12, 5))

for i, ax in enumerate(axes.flat):
    ax.imshow(x_train_full[i], cmap="gray")
    ax.set_title(
        f"Label: {class_names[y_train_full[i]]}"
    )
    ax.axis("off")

plt.suptitle("Sample Fashion-MNIST Images")
plt.tight_layout()

sample_path = os.path.join(
    OUTPUT_DIR, "01_sample_images.png"
)
plt.savefig(sample_path, dpi=150, bbox_inches="tight")
plt.show()

print("Saved:", sample_path)
print()


# ================================================================
# STEP 1B: CLASS DISTRIBUTION
# ================================================================

print("=" * 70)
print("STEP 1B: CLASS DISTRIBUTION")
print("=" * 70)

class_counts = np.bincount(
    y_train_full, minlength=NUM_CLASSES
)

for i, count in enumerate(class_counts):
    print(f"{i}: {class_names[i]:15s} -> {count} images")

plt.figure(figsize=(11, 5))
plt.bar(class_names, class_counts)
plt.title("Fashion-MNIST Training Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Images")
plt.xticks(rotation=40, ha="right")
plt.tight_layout()

distribution_path = os.path.join(
    OUTPUT_DIR, "02_class_distribution.png"
)
plt.savefig(distribution_path, dpi=150, bbox_inches="tight")
plt.show()

print("Saved:", distribution_path)
print()


# ================================================================
# STEP 2: IMAGE PREPROCESSING USING PILLOW
# ================================================================

print("=" * 70)
print("STEP 2: IMAGE PREPROCESSING USING PILLOW")
print("=" * 70)

sample_array = x_train_full[0]

sample_image = Image.fromarray(
    sample_array.astype(np.uint8)
)

print("Original image size:", sample_image.size)
print("Original image mode:", sample_image.mode)
print("Image category:", class_names[y_train_full[0]])

# Operation 1: Resize
resized_image = sample_image.resize((56, 56))

# Operation 2: Convert grayscale to RGB
rgb_image = sample_image.convert("RGB")

# Operation 3: Enhance contrast
enhanced_image = ImageEnhance.Contrast(
    sample_image
).enhance(1.5)

# Operation 4: Gaussian smoothing
smoothed_image = sample_image.filter(
    ImageFilter.GaussianBlur(radius=0.7)
)

preprocessed_images = [
    sample_image,
    resized_image,
    rgb_image,
    enhanced_image,
    smoothed_image
]

preprocessing_titles = [
    "Original grayscale",
    "Resized 56 x 56",
    "RGB conversion",
    "Enhanced contrast",
    "Gaussian smoothing"
]

fig, axes = plt.subplots(1, 5, figsize=(16, 4))

for ax, image, title in zip(
    axes, preprocessed_images, preprocessing_titles
):
    if image.mode == "RGB":
        ax.imshow(image)
    else:
        ax.imshow(image, cmap="gray")

    ax.set_title(title)
    ax.axis("off")

plt.suptitle("Pillow Image Preprocessing Operations")
plt.tight_layout()

preprocessing_path = os.path.join(
    OUTPUT_DIR, "03_preprocessing.png"
)
plt.savefig(preprocessing_path, dpi=150, bbox_inches="tight")
plt.show()

print("Saved:", preprocessing_path)
print()


# ================================================================
# PIXEL VALUE INSPECTION
# ================================================================

print("=" * 70)
print("PIXEL VALUE INSPECTION")
print("=" * 70)

pixel_array = np.asarray(sample_image)

print("Pixel array shape:", pixel_array.shape)
print("Minimum original intensity:", pixel_array.min())
print("Maximum original intensity:", pixel_array.max())
print("Mean original intensity:", pixel_array.mean())
print()


# ================================================================
# STEP 3: IMAGE PROCESSING WITH NUMPY AND PILLOW
# ================================================================

print("=" * 70)
print("STEP 3: IMAGE PROCESSING")
print("=" * 70)

gray_array = np.asarray(sample_image)

# Operation 1: Binary thresholding
threshold = 100

binary_array = np.where(
    gray_array >= threshold, 255, 0
).astype(np.uint8)

binary_image = Image.fromarray(binary_array)

# Operation 2: Edge detection
edge_image = sample_image.filter(ImageFilter.FIND_EDGES)

# Operation 3: Gaussian smoothing
smooth_for_comparison = sample_image.filter(
    ImageFilter.GaussianBlur(radius=1.0)
)

processing_images = [
    sample_image,
    binary_image,
    edge_image,
    smooth_for_comparison
]

processing_titles = [
    "Original image",
    f"Threshold = {threshold}",
    "Edge detection",
    "Gaussian smoothing"
]

fig, axes = plt.subplots(1, 4, figsize=(13, 4))

for ax, image, title in zip(
    axes, processing_images, processing_titles
):
    ax.imshow(image, cmap="gray")
    ax.set_title(title)
    ax.axis("off")

plt.suptitle("Image Processing Techniques")
plt.tight_layout()

processing_path = os.path.join(
    OUTPUT_DIR, "04_image_processing.png"
)
plt.savefig(processing_path, dpi=150, bbox_inches="tight")
plt.show()

print("Saved:", processing_path)
print()


# ================================================================
# STEP 3A: EFFECT OF DIFFERENT THRESHOLD VALUES
# ================================================================

print("=" * 70)
print("STEP 3A: THRESHOLD COMPARISON")
print("=" * 70)

thresholds = [50, 100, 150, 200]

fig, axes = plt.subplots(1, 4, figsize=(12, 3))

for ax, threshold_value in zip(axes, thresholds):
    result = np.where(
        gray_array >= threshold_value, 255, 0
    ).astype(np.uint8)

    ax.imshow(result, cmap="gray")
    ax.set_title(f"Threshold = {threshold_value}")
    ax.axis("off")

plt.suptitle("Effect of Different Threshold Values")
plt.tight_layout()

threshold_path = os.path.join(
    OUTPUT_DIR, "05_threshold_comparison.png"
)
plt.savefig(threshold_path, dpi=150, bbox_inches="tight")
plt.show()

print("Saved:", threshold_path)
print()


# ================================================================
# STEP 4: PREPARE TRAINING, VALIDATION AND TEST DATA
# ================================================================

print("=" * 70)
print("STEP 4: PREPARE TRAINING, VALIDATION AND TEST DATA")
print("=" * 70)

# Fashion-MNIST contains:
# 60,000 original training images
# 10,000 independent test images
#
# Split the original training data into:
# 48,000 training images
# 12,000 validation images
#
# The independent test dataset is kept untouched.

train_size = 48000

x_train_raw = x_train_full[:train_size]
y_train = y_train_full[:train_size]

x_val_raw = x_train_full[train_size:]
y_val = y_train_full[train_size:]

print("Training images:", len(x_train_raw))
print("Validation images:", len(x_val_raw))
print("Test images:", len(x_test_raw))
print()


# ================================================================
# STEP 4A: NORMALIZE PIXELS AND ADD CHANNEL DIMENSION
# ================================================================

print("=" * 70)
print("STEP 4A: NORMALIZE PIXELS AND PREPARE CNN INPUTS")
print("=" * 70)

# CNN input shape: (height, width, channels)
# Fashion-MNIST images are grayscale, so channels = 1.
#
# Normalize integer pixels from [0, 255] to [0, 1].

x_train = x_train_raw.astype("float32") / 255.0
x_val = x_val_raw.astype("float32") / 255.0
x_test = x_test_raw.astype("float32") / 255.0

x_train = np.expand_dims(x_train, axis=-1)
x_val = np.expand_dims(x_val, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)

print("Training shape:", x_train.shape)
print("Validation shape:", x_val.shape)
print("Test shape:", x_test.shape)
print("Training minimum:", x_train.min())
print("Training maximum:", x_train.max())
print()


# ================================================================
# STEP 5: BUILD THE CONVOLUTIONAL NEURAL NETWORK
# ================================================================

print("=" * 70)
print("STEP 5: BUILD THE CONVOLUTIONAL NEURAL NETWORK")
print("=" * 70)

# CNN architecture:
#
# Input: 28 x 28 x 1
#
# Conv2D: 32 filters, 3 x 3, ReLU
# MaxPooling2D: 2 x 2
#
# Conv2D: 64 filters, 3 x 3, ReLU
# MaxPooling2D: 2 x 2
#
# Flatten
# Dense: 64 neurons, ReLU
# Dense: 10 output logits
#
# SparseCategoricalCrossentropy(from_logits=True)
# is used during compilation.

model = keras.Sequential(
    [
        layers.Input(shape=(28, 28, 1)),

        layers.Conv2D(
            32,
            kernel_size=(3, 3),
            padding="same",
            activation="relu",
            name="conv2d_1"
        ),

        layers.MaxPooling2D(
            pool_size=(2, 2),
            name="max_pooling_1"
        ),

        layers.Conv2D(
            64,
            kernel_size=(3, 3),
            padding="same",
            activation="relu",
            name="conv2d_2"
        ),

        layers.MaxPooling2D(
            pool_size=(2, 2),
            name="max_pooling_2"
        ),

        layers.Flatten(name="flatten"),

        layers.Dense(
            64,
            activation="relu",
            name="dense_features"
        ),

        layers.Dense(
            NUM_CLASSES,
            name="class_logits"
        )
    ],
    name="FashionMNIST_CNN"
)

# Display model architecture
model.summary()

print()


# ================================================================
# STEP 5A: COUNT MODEL PARAMETERS
# ================================================================

total_parameters = model.count_params()

print("Total trainable parameters:", total_parameters)
print()


# ================================================================
# STEP 5B: COMPILE THE CNN
# ================================================================

criterion = keras.losses.SparseCategoricalCrossentropy(
    from_logits=True
)

optimizer = keras.optimizers.Adam(
    learning_rate=LEARNING_RATE
)

model.compile(
    optimizer=optimizer,
    loss=criterion,
    metrics=["accuracy"]
)

print("CNN compiled successfully.")
print()


# ================================================================
# STEP 6: TRAIN THE CNN
# ================================================================

print("=" * 70)
print("STEP 6: TRAIN THE CNN")
print("=" * 70)

history = model.fit(
    x_train,
    y_train,
    validation_data=(x_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    shuffle=True,
    verbose=1
)

print()


# ================================================================
# STEP 6A: EXTRACT TRAINING HISTORY
# ================================================================

train_losses = history.history["loss"]
val_losses = history.history["val_loss"]

train_accuracies = history.history["accuracy"]
val_accuracies = history.history["val_accuracy"]


# ================================================================
# STEP 6B: PLOT LEARNING CURVES
# ================================================================

print("=" * 70)
print("STEP 6B: PLOT TRAINING AND VALIDATION CURVES")
print("=" * 70)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Loss curves
axes[0].plot(
    range(1, EPOCHS + 1),
    train_losses,
    marker="o",
    label="Training loss"
)

axes[0].plot(
    range(1, EPOCHS + 1),
    val_losses,
    marker="o",
    label="Validation loss"
)

axes[0].set_title("Loss Across Epochs")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Cross-entropy loss")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Accuracy curves
axes[1].plot(
    range(1, EPOCHS + 1),
    train_accuracies,
    marker="o",
    label="Training accuracy"
)

axes[1].plot(
    range(1, EPOCHS + 1),
    val_accuracies,
    marker="o",
    label="Validation accuracy"
)

axes[1].set_title("Accuracy Across Epochs")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Accuracy")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()

learning_curve_path = os.path.join(
    OUTPUT_DIR, "06_learning_curves.png"
)
plt.savefig(learning_curve_path, dpi=150, bbox_inches="tight")
plt.show()

print("Saved:", learning_curve_path)
print()


# ================================================================
# STEP 7: FINAL EVALUATION ON UNSEEN TEST IMAGES
# ================================================================

print("=" * 70)
print("STEP 7: FINAL TEST EVALUATION")
print("=" * 70)

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    batch_size=256,
    verbose=0
)

print(f"Test loss: {test_loss:.4f}")
print(f"Test accuracy: {test_accuracy:.2%}")
print()


# ================================================================
# STEP 7A: GENERATE TEST PREDICTIONS
# ================================================================

print("=" * 70)
print("STEP 7A: GENERATE TEST PREDICTIONS")
print("=" * 70)

# The model returns logits, not probabilities.
logits = model.predict(
    x_test,
    batch_size=256,
    verbose=0
)

all_predictions = np.argmax(logits, axis=1)
all_labels = y_test

test_accuracy_sklearn = accuracy_score(
    all_labels, all_predictions
)

print(
    "Accuracy from scikit-learn:",
    round(test_accuracy_sklearn, 4)
)
print()


# ================================================================
# STEP 7B: CLASSIFICATION REPORT
# ================================================================

print("=" * 70)
print("STEP 7B: CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        all_labels,
        all_predictions,
        labels=list(range(NUM_CLASSES)),
        target_names=class_names,
        digits=3,
        zero_division=0
    )
)


# ================================================================
# STEP 7C: CONFUSION MATRIX
# ================================================================

print("=" * 70)
print("STEP 7C: CONFUSION MATRIX")
print("=" * 70)

cm = confusion_matrix(
    all_labels,
    all_predictions,
    labels=list(range(NUM_CLASSES))
)

fig, ax = plt.subplots(figsize=(11, 8))

image_plot = ax.imshow(cm, cmap="Blues")
fig.colorbar(image_plot, ax=ax)

ax.set_xticks(range(NUM_CLASSES))
ax.set_yticks(range(NUM_CLASSES))

ax.set_xticklabels(
    class_names, rotation=45, ha="right"
)
ax.set_yticklabels(class_names)

ax.set_xlabel("Predicted class")
ax.set_ylabel("Actual class")
ax.set_title("Fashion-MNIST CNN Test Confusion Matrix")

for row in range(cm.shape[0]):
    for col in range(cm.shape[1]):
        text_color = (
            "white"
            if cm[row, col] > cm.max() * 0.5
            else "black"
        )

        ax.text(
            col,
            row,
            str(cm[row, col]),
            ha="center",
            va="center",
            fontsize=7,
            color=text_color
        )

plt.tight_layout()

confusion_matrix_path = os.path.join(
    OUTPUT_DIR, "07_confusion_matrix.png"
)
plt.savefig(
    confusion_matrix_path, dpi=150, bbox_inches="tight"
)
plt.show()

print("Saved:", confusion_matrix_path)
print()


# ================================================================
# STEP 7D: MOST FREQUENT CLASS CONFUSIONS
# ================================================================

print("=" * 70)
print("STEP 7D: MOST FREQUENT CLASS CONFUSIONS")
print("=" * 70)

confusion_without_diagonal = cm.copy()
np.fill_diagonal(confusion_without_diagonal, 0)

print("Top class confusions:")

for _ in range(5):
    actual_idx, predicted_idx = np.unravel_index(
        np.argmax(confusion_without_diagonal),
        confusion_without_diagonal.shape
    )

    confusion_count = confusion_without_diagonal[
        actual_idx, predicted_idx
    ]

    if confusion_count == 0:
        break

    print(
        f"Actual: {class_names[actual_idx]} | "
        f"Predicted: {class_names[predicted_idx]} | "
        f"Count: {confusion_count}"
    )

    confusion_without_diagonal[
        actual_idx, predicted_idx
    ] = 0

print()


# ================================================================
# STEP 8: VISUALIZE INCORRECT PREDICTIONS
# ================================================================

print("=" * 70)
print("STEP 8: ANALYZE INCORRECT PREDICTIONS")
print("=" * 70)

incorrect_indices = np.where(
    all_predictions != all_labels
)[0]

print("Total incorrect predictions:", len(incorrect_indices))
print(
    "Total correct predictions:",
    len(all_labels) - len(incorrect_indices)
)
print()

number_of_errors_to_plot = min(
    10, len(incorrect_indices)
)

selected_errors = incorrect_indices[
    :number_of_errors_to_plot
]

if number_of_errors_to_plot > 0:
    columns = min(5, number_of_errors_to_plot)
    rows = (
        number_of_errors_to_plot + columns - 1
    ) // columns

    fig, axes = plt.subplots(
        rows,
        columns,
        figsize=(3 * columns, 3 * rows),
        squeeze=False
    )

    axes = axes.flatten()

    for i, index in enumerate(selected_errors):
        axes[i].imshow(
            x_test[index].squeeze(),
            cmap="gray"
        )

        actual = class_names[all_labels[index]]
        predicted = class_names[all_predictions[index]]

        axes[i].set_title(
            f"Actual: {actual}\nPredicted: {predicted}",
            fontsize=9
        )
        axes[i].axis("off")

    for i in range(number_of_errors_to_plot, len(axes)):
        axes[i].axis("off")

    plt.suptitle("Examples of Incorrect CNN Predictions")
    plt.tight_layout()

    misclassified_path = os.path.join(
        OUTPUT_DIR, "08_misclassified_images.png"
    )
    plt.savefig(
        misclassified_path, dpi=150, bbox_inches="tight"
    )
    plt.show()

    print("Saved:", misclassified_path)
else:
    print("No misclassified images were found.")

print(
    "Discussion: Examine visually similar classes, "
    "low image resolution, and ambiguous shapes."
)
print()


# ================================================================
# STEP 9: PREDICT THE CATEGORY OF A NEW IMAGE
# ================================================================

print("=" * 70)
print("STEP 9: PREDICT A NEW IMAGE")
print("=" * 70)

# Use a held-out test image as a simulated new image.
new_image = x_test[0]
actual_label = y_test[0]

input_image = np.expand_dims(new_image, axis=0)

new_logits = model.predict(input_image, verbose=0)

probabilities = tf.nn.softmax(
    new_logits, axis=1
).numpy()

predicted_class = int(np.argmax(probabilities[0]))
confidence = float(probabilities[0, predicted_class])

print("Predicted category:", class_names[predicted_class])
print("Actual category:", class_names[actual_label])
print(f"Model confidence: {confidence:.2%}")
print()

plt.figure(figsize=(5, 5))
plt.imshow(new_image.squeeze(), cmap="gray")

plt.title(
    f"Prediction: {class_names[predicted_class]}\n"
    f"Confidence: {confidence:.2%}\n"
    f"Actual: {class_names[actual_label]}"
)

plt.axis("off")
plt.tight_layout()

new_prediction_path = os.path.join(
    OUTPUT_DIR, "09_new_image_prediction.png"
)
plt.savefig(
    new_prediction_path, dpi=150, bbox_inches="tight"
)
plt.show()

print("Saved:", new_prediction_path)
print()

print(
    "IMPORTANT: Softmax confidence is not necessarily "
    "a calibrated probability that a prediction is correct."
)
print()


# ================================================================
# STEP 9A: PREDICT AN EXTERNAL GRAYSCALE IMAGE
# ================================================================

print("=" * 70)
print("STEP 9A: EXTERNAL IMAGE PREDICTION FUNCTION")
print("=" * 70)


def predict_external_image(image_path, model):
    """
    Load a local image, convert it to grayscale, resize it
    to 28 x 28, normalize its pixels to [0, 1], add the
    CNN channel and batch dimensions, and predict a class.

    This basic pipeline is intended for images visually
    similar to Fashion-MNIST.

    Arbitrary factory photographs may not be classified
    reliably by this model.
    """

    # Load image and convert to grayscale.
    external_image = Image.open(image_path).convert("L")

    # Resize to Fashion-MNIST dimensions.
    external_image = external_image.resize((28, 28))

    # Convert to floating-point NumPy array.
    image_array = np.asarray(
        external_image, dtype=np.float32
    )

    # Normalize pixels.
    image_array = image_array / 255.0

    # Add channel dimension: (28, 28, 1).
    image_array = np.expand_dims(image_array, axis=-1)

    # Add batch dimension: (1, 28, 28, 1).
    image_tensor = np.expand_dims(image_array, axis=0)

    # Predict logits and convert to probabilities.
    external_logits = model.predict(
        image_tensor, verbose=0
    )

    external_probabilities = tf.nn.softmax(
        external_logits, axis=1
    ).numpy()

    predicted_idx = int(
        np.argmax(external_probabilities[0])
    )

    confidence_value = float(
        external_probabilities[0, predicted_idx]
    )

    print(
        "External image prediction:",
        class_names[predicted_idx]
    )
    print(f"Model confidence: {confidence_value:.2%}")

    plt.figure(figsize=(4, 4))
    plt.imshow(external_image, cmap="gray")
    plt.title(
        f"{class_names[predicted_idx]}\n"
        f"Confidence: {confidence_value:.2%}"
    )
    plt.axis("off")
    plt.tight_layout()
    plt.show()

    return class_names[predicted_idx], confidence_value


print("External image prediction function created.")
print()

print("To classify an external image in Colab:")
print("1. Upload a compatible image.")
print("2. Run the following command:")
print()
print(
    "predicted_name, confidence_value = "
    "predict_external_image('my_product_image.png', model)"
)
print()


# ================================================================
# STEP 10: SAVE THE TRAINED CNN
# ================================================================

print("=" * 70)
print("STEP 10: SAVE THE TRAINED CNN")
print("=" * 70)

model_path = os.path.join(
    OUTPUT_DIR, "fashion_cnn_classifier.keras"
)

model.save(model_path)

print("Model saved to:", model_path)
print()


# ================================================================
# STEP 10A: SAVE MODEL METADATA
# ================================================================

metadata_path = os.path.join(
    OUTPUT_DIR, "model_metadata.txt"
)

with open(metadata_path, "w", encoding="utf-8") as file:
    file.write("AAI-521 Fashion-MNIST CNN Classifier\n")
    file.write(f"TensorFlow version: {tf.__version__}\n")
    file.write(f"Keras version: {keras.__version__}\n")
    file.write(f"Epochs: {EPOCHS}\n")
    file.write(f"Batch size: {BATCH_SIZE}\n")
    file.write(f"Learning rate: {LEARNING_RATE}\n")
    file.write("Architecture: CNN\n")
    file.write("Input size: 28 x 28 x 1 grayscale\n")
    file.write("Input normalization: pixel / 255.0\n")
    file.write("Convolution filters: 32 and 64\n")
    file.write("Pooling: 2 x 2 MaxPooling2D\n")
    file.write("Dense hidden layer: 64 ReLU units\n")
    file.write("Output: 10 logits\n")
    file.write(
        f"Final training accuracy: "
        f"{train_accuracies[-1]:.6f}\n"
    )
    file.write(
        f"Final validation accuracy: "
        f"{val_accuracies[-1]:.6f}\n"
    )
    file.write(f"Final test accuracy: {test_accuracy:.6f}\n")
    file.write(f"Final test loss: {test_loss:.6f}\n")
    file.write(f"Random seed: {SEED}\n")
    file.write(f"Class names: {class_names}\n")
    file.write(
        "Limitation: Fashion-MNIST classifies clothing; "
        "it does not identify manufacturing defects.\n"
    )

print("Metadata saved to:", metadata_path)
print()


# ================================================================
# STEP 10B: RELOAD THE SAVED CNN
# ================================================================

print("=" * 70)
print("STEP 10B: RELOAD THE SAVED CNN")
print("=" * 70)

loaded_model = keras.models.load_model(model_path)

print("Model reloaded successfully.")

original_prediction = model.predict(
    input_image, verbose=0
)

reloaded_prediction = loaded_model.predict(
    input_image, verbose=0
)

predictions_match = np.allclose(
    original_prediction,
    reloaded_prediction,
    atol=1e-6
)

print(
    "Original and reloaded outputs match:",
    predictions_match
)
print()


# ================================================================
# STEP 11: EXPERIMENT SUMMARY
# ================================================================

print("=" * 70)
print("STEP 11: EXPERIMENT SUMMARY")
print("=" * 70)

print("Model architecture: Convolutional Neural Network (CNN)")
print("Input shape: 28 x 28 x 1")
print("Convolution layers: 32 and 64 filters")
print("Pooling layers: Two 2 x 2 MaxPooling2D layers")
print("Dense hidden layer: 64 ReLU neurons")
print("Output classes:", NUM_CLASSES)
print("Optimizer: Adam")
print("Loss: SparseCategoricalCrossentropy from logits")
print("Learning rate:", LEARNING_RATE)
print("Epochs:", EPOCHS)
print(f"Final training accuracy: {train_accuracies[-1]:.2%}")
print(f"Final validation accuracy: {val_accuracies[-1]:.2%}")
print(f"Final test accuracy: {test_accuracy:.2%}")
print(f"Final test loss: {test_loss:.4f}")
print()


# ================================================================
# STEP 11A: DATASET SUMMARY
# ================================================================

print("=" * 70)
print("DATASET SUMMARY")
print("=" * 70)

print("Training images: 48,000")
print("Validation images: 12,000")
print("Test images: 10,000")
print("Image size: 28 x 28 pixels")
print("Image channels: 1 (grayscale)")
print("Number of classes: 10")
print()


# ================================================================
# STEP 11B: SAVED OUTPUT FILES
# ================================================================

print("=" * 70)
print("SAVED OUTPUT FILES")
print("=" * 70)

for filename in sorted(os.listdir(OUTPUT_DIR)):
    print(" -", os.path.join(OUTPUT_DIR, filename))

print()


# ================================================================
# STEP 12: IMPROVEMENT EXPERIMENTS FOR STUDENTS
# ================================================================

print("=" * 70)
print("STEP 12: SUGGESTED IMPROVEMENT EXPERIMENTS")
print("=" * 70)

print(
    """
Experiment A: Learning Rate
    Compare:
        0.01
        0.001
        0.0001

Experiment B: Architecture
    Compare:
        CNN with 32 and 64 filters
        Smaller CNN
        Larger CNN
        CNN with an additional convolutional block

Experiment C: Data Augmentation
    Test:
        Small rotations
        Small translations
        Horizontal flips only where class meaning is preserved

Experiment D: Class-Level Evaluation
    Compare:
        Precision
        Recall
        F1-score
    for all 10 classes.

Experiment E: Error Analysis
    Examine:
        Confusion matrix
        Incorrectly classified images
        Visually similar classes

Experiment F: Regularization
    Experiment with:
        Dropout
        L2 regularization
        Batch normalization

Experiment G: CNN Architecture
    Compare different filter counts, kernel sizes and pooling.

Experiment H: Compare CNN against MLP
    Train both models with the same training and validation split.
    Compare parameter counts, accuracy, training time and errors.

Important experimental principle:
    Use the same training/validation split when comparing models.
    Select the final configuration using validation results.
    Use the independent test dataset for final evaluation.
    Avoid repeatedly tuning the model on the test dataset.
"""
)

print()


# ================================================================
# FINAL CASE STUDY MESSAGE
# ================================================================

print("=" * 70)
print("CASE STUDY COMPLETED SUCCESSFULLY")
print("=" * 70)

print()
print("Educational limitation:")
print("This CNN categorizes Fashion-MNIST clothing images.")
print("It does NOT detect manufacturing defects.")
print()
print("Real manufacturing quality inspection requires:")
print("  1. Representative industrial images")
print("  2. Appropriate defect/non-defect labels")
print("  3. Multiple relevant defect categories")
print("  4. Validation using actual factory images")
print("  5. Testing under real operating conditions")
print()
print("All preprocessing, training, evaluation and saving")
print("steps have been completed.")
print()
print("Output directory:", OUTPUT_DIR)
print("=" * 70)
```
